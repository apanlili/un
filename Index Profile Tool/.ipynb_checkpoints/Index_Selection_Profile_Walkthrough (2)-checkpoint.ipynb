# The Index Selection Profile — a walkthrough

This notebook is the tool explained by running it. Every cell can be executed;
nothing here is a description of code that lives elsewhere.

**What the tool answers.** A ministry is deciding whether to adopt a development
index — to put it in a plan, set a target on it, and report against it for
years. That decision is usually made on the index's reputation. This tool asks
the questions that should be asked instead, and answers them from the
publisher's own files.

**What it refuses to do.** It never produces an overall suitability score. A
decision is weighed, not averaged, and a single number would hide the trade-offs
the choice turns on.

---

One environment variable, pointing at the folder holding the index workbooks.

````python
import os, sys
from pathlib import Path

# --- the only two things to set -------------------------------------------
# 1. The folder CONTAINING the isf/ directory.
#    If you launched Jupyter from inside it, leave this as Path(".").
PACKAGE = Path(".")

# 2. The folder holding the index metadata and calculation workbooks.
DATA = Path("/mnt/user-data/uploads")      # <- change this
# ---------------------------------------------------------------------------

sys.path.insert(0, str(PACKAGE.resolve()))
os.environ["ISF_DATA_DIR"] = str(DATA.resolve())

from isf.engine_core import assess, RULEBOOK, REGISTRY, TIER3_PROMPTS

print(f"reading index files from: {DATA.resolve()}")
print(f"{len(REGISTRY)} indices registered")
print(f"{len(RULEBOOK)} criteria graded by us")
print(f"{len(TIER3_PROMPTS)} questions left to the policymaker")
````

## 1 · The questions the tool asks

Fourteen criteria. Three of them are **gates** — a failure there is decisive in a
particular way, explained in section 3. The rest describe how well the index is
built and how it behaves over time.

Run this to see all of them, with the question each one puts.

````python
for key, rule in RULEBOOK.items():
    gate = "  [GATE]" if rule["gate"] else ""
    print(f"{rule['num']:>2}. {rule['name']}{gate}")
    print(f"    {rule['question']}")
````

### And four questions the tool does not answer

These are about your strategy, not about the index. An instrument built
elsewhere should not overrule a judgment that belongs here, so the tool poses
them and stops.

````python
for prompt in TIER3_PROMPTS:
    print(f"• {prompt['name']}: {prompt['question']}")
````

## 2 · Assessing the corpus

`assess()` reads every index's files and grades it. This takes a minute or so —
it is opening thirty workbooks.

````python
profiles = assess()
print(f"{len(profiles)} indices assessed")
print(f"{len(profiles[0]['rows'])} criteria each")
````

### One index, read end to end

The output is a **profile**: a column of gestures, not a score. Green means
clear, yellow means this is a judgment call, red means stop and check.

`awaiting` means we have not assessed it yet, and `not assessable` means the
publisher does not release what the check needs. Neither is a pass — the tool
never invents a grade to complete a picture.

````python
MARK = {"green": "clear        ", "yellow": "judgment call",
        "red": "STOP & CHECK ", "awaiting": "not yet      ",
        "not_assessable": "cannot check ", "na": "n/a          "}

def show(index_id):
    profile = next(p for p in profiles if p["index_id"] == index_id)
    print(profile["index_name"], "—", profile.get("publisher", ""))
    print("=" * 78)
    for row in profile["rows"]:
        gate = " (gate)" if row["gate"] else ""
        print(f"  {MARK[row['status']]}  {row['name']}{gate}")
    print()
    print("VERDICT:", profile["verdict"])

show("EGDI")
````

---

## 3 · What a failed gate actually means

This is the part most worth understanding, because it changed and it is the
whole point of the tool.

A gate failure does **not** mean the index is bad. It means the index cannot
carry certain *uses*. There are four things a government adopts an index to do,
and an index can be sound for some and unsafe for others.

````python
from isf.engine_core.rulebook import PURPOSES, PURPOSE_MAP

for key, label in PURPOSES.items():
    print(f"• {label}")

print()
print("Which criteria can block which purpose:")
for key, rule in RULEBOOK.items():
    print(f"  {rule['name']:<28} blocks: {', '.join(PURPOSE_MAP[key])}")
````

So the verdict names what the index **can** do and what it **cannot**, rather
than passing or failing it.

An index whose raw data is not published cannot be used to set a target — you
could not defend the number. It may be perfectly good for visibility. Saying so
is what stops the tool telling a government its flagship index is worthless.

````python
for profile in profiles:
    if profile["gate_failures"]:
        print(f"{profile['index_id']:<7} {profile['verdict']}")
        print()
````

---

## 4 · The same index reads differently for different countries

Five criteria depend on **who is asking**. How much of *your* score is estimated
rather than measured, how old *your* data is, whether *your* position sits in a
crowded band — these are facts about a country, not about an index.

Where a country grade differs from the overall one, the tool keeps both and says
which is which. That difference is a finding, not an inconsistency.

````python
from isf.engine import load_engine
from isf.metadata_adapter import metadata_rows
from isf import country_layer

engine = load_engine()
rows = metadata_rows(engine, "Global Innovation Index")

for country in ["Jordan", "Egypt", "Somalia"]:
    result = country_layer.provenance(rows, country)
    print(f"{country}: {result.grade.upper()}")
    print(f"   {result.note}")
    print()
````

Notice that the sentence and the grade agree. They come from the same
calculation — earlier they were computed separately and could contradict each
other, which is the fastest way to lose a reader's trust.

---

## 5 · The guard that makes the grades trustworthy

The strongest claim the tool makes: for every index with a verified structure,
it re-reads the publisher's own file and recomputes **their** published headline
score. If it cannot reproduce their number, something has changed and the grades
cannot be trusted until it is understood.

````python
from isf.engine_core import run_self_check

summary = run_self_check(verbose=False)
print(summary)
print()
print("Everything reproduces." if summary.ok else "SOMETHING FAILED:")
for failure in summary.failures:
    print(" ", failure["id"], failure["note"])
````

### Proving the guard actually works

A check that cannot fail is worse than no check. This deliberately breaks one
index's structure — as if a publisher had moved a column — and confirms the
guard catches it.

````python
from isf.engine_core import VERIFIED_STRUCTURES

spec = VERIFIED_STRUCTURES["HDI"]
original = spec["final_col"]

spec["final_col"] = original - 1          # pretend a column moved
broken = run_self_check(verbose=False)
print("with a moved column:", broken)

spec["final_col"] = original              # put it back
print("restored:          ", run_self_check(verbose=False))
````

## 6 · One finding, in full

The clearest example of what the tool sees that a reputation does not.

The Human Development Index scores income on a scale that stops at $75,000.
Qatar earns about $105,000, so its income score is pinned at the maximum: it
could lose a large share of its income and its HDI would not move. What actually
moves Qatar's position is schooling.

So for Qatar, a development index behaves like a measure of human capital.

````python
import openpyxl
from isf.engine_core import DATA_DIR, GOALPOST_INDICES
from isf.engine_core.goalpost import goalpost_headroom, goalpost_note

spec = VERIFIED_STRUCTURES["HDI"]
workbook = openpyxl.load_workbook(DATA_DIR / spec["file"], read_only=True, data_only=True)
sheet_rows = list(workbook[spec["sheet"]].iter_rows(values_only=True))

headroom = goalpost_headroom("HDI", sheet_rows, spec["country_col"],
                             spec["data_start"], skip=spec.get("skip", []))

print(f"{headroom['n_countries']} countries")
print(f"{headroom['n_frozen_any']} have at least one dimension frozen at its ceiling")
print(f"{headroom['n_saturated_any']} have at least one near it")
print()
for country in ["Qatar", "Norway", "Ireland", "Jordan"]:
    record = headroom["per_country"].get(country)
    if record:
        print(f"{country:<10} can still move on {record['effective_dimensionality']} "
              f"of {record['n_dimensions']} dimensions   frozen: {record['frozen'] or 'none'}")
````

Note Norway and Ireland: **more** constrained than Qatar. This is not a Gulf
phenomenon — the most frequently frozen dimension across the whole index is
expected years of schooling, and the countries it affects are mostly European.

That matters for how the finding is presented: it is a property of how the index
is built, not a story about oil.

````python
print(goalpost_note(headroom, "Qatar"))
````

## 7 · Producing the outputs

Two commands. The first reads the engine and writes a workbook; the second turns
that workbook into the card deck.

The workbook is the join between them. **Nothing downstream needs the engine** —
which is why the deck can be shared without giving away the instrument.

````python
# In a terminal, from the package folder:
#
#   export ISF_DATA_DIR=/path/to/index/files
#   python -m isf.build_export     # engine   -> index_cards_export.xlsx
#   python -m isf.build_deck       # workbook -> index_flipcards.html
#
# Or from here:
from isf.build_export import build

report = build()
print(report.summary())
````

### Editing the wording without touching code

The four columns a reader actually sees — the question, the answer, the action
line, and the technical explanation — are **columns in that workbook**.

Edit them in Excel, re-run `python -m isf.build_deck`, and the deck shows your
wording. No Python involved.

````python
import openpyxl

sheet = openpyxl.load_workbook(report.output if hasattr(report, "output")
                               else "/mnt/user-data/outputs/index_cards_export.xlsx",
                               data_only=True)["Cards"]
headers = [sheet.cell(1, c).value for c in range(1, sheet.max_column + 1)]

print("Editable text columns:")
for column in ["Question (card front)", "Short answer (card back)",
               "What to do", "Hover / more (technical reason)"]:
    print(f"  column {headers.index(column) + 1}: {column}")

print()
print("Example row:")
for column in ["Criterion", "Question (card front)", "Short answer (card back)"]:
    print(f"  {column}: {sheet.cell(2, headers.index(column) + 1).value}")
````

## 9 · What the tool does not do

Stated plainly, because a tool that hides its limits is worse than one that has
none.

**Five criteria are read, not computed.** Responsiveness, faithfulness,
weighting, stability and comparability come from reading each publisher's
methodology. Each carries a citation, and the Confidence Register records how
well each is sourced — but no outside reviewer has checked them. This is the
largest open item.

**A methodology change is not detected automatically.** The self-check catches a
publisher moving a *column*. Nothing catches a publisher changing its *method* —
that requires someone re-reading the documentation. It is a scheduled human
task, and it is how the four methodology breaks in the current corpus were
found.

**Country-level findings cover the Arab states.** The metadata is a regional
slice, so currency and provenance are computed for those countries. Calculation
findings cover the index's full country set.

**It has never been used in a live selection decision.** Its presentation, and
possibly its structure, should be expected to change once it has.

````python
from isf.audit import LIMITATIONS

for title, meaning, remedy in LIMITATIONS:
    print(f"• {title}")
    print(f"    {meaning}")
    print(f"    → {remedy}")
    print()
````

## Where things live

| File | What it is |
|---|---|
| `isf/engine_core/rulebook.py` | The criteria, thresholds, purposes. **Read this one first.** |
| `isf/engine_core/corpus.py` | Which indices, and the evidence behind each reviewed grade |
| `isf/engine_core/selfcheck.py` | The reproduction guard |
| `isf/wording.py` | Default card text |
| `isf/build_export.py` | Engine → workbook |
| `isf/build_deck.py` | Workbook → deck |

To add an index: one line in `REGISTRY` in `corpus.py`. Construction criteria
compute straight away; reviewed criteria read `awaiting` until someone reads the
methodology.

## 10 · Plugging in new files

The part most likely to be needed next.

**Adding an index.** Put its files in the data folder, then add one line to
`REGISTRY` in `isf/engine_core/corpus.py`:

```python
"WJP Rule of Law": ("Rule_of_Law_metadata.xlsx", "Rule_of_Law_calc.xlsx"),
```

Re-run the export. Country-level and construction findings compute at once. The
five read-from-methodology criteria show `awaiting` until somebody reads the
publisher's documentation — the tool will not invent them.

Only once the self-check reproduces the published score should a
`VERIFIED_STRUCTURES` entry be added. That entry is a claim the guard checks on
every run, so it should not be made before it is true.

**What the metadata file must contain.** One row per indicator per country, with
these column headings. They are matched by name, so column order does not
matter.

````python
from isf.engine_core.country_lens import _meta_country_rows
from isf.engine_core import DATA_DIR, REGISTRY
import openpyxl

# The shape a metadata file must have, taken from one that works.
example = DATA_DIR / REGISTRY["Global Innovation Index"][0]
headers = [c.value for c in next(openpyxl.load_workbook(
    example, read_only=True).worksheets[0].iter_rows(max_row=1))]

print("Columns the parser looks for:")
for h in headers[:10]:
    if h:
        print(f"  {h}")
````

**Known gaps in the current files** — data problems, not code problems:

- **GovTech** is missing seven indicators. The publisher's sheet has 201; the
  metadata has 194, identically for all 22 countries, in adjacent columns —
  one skipped block during transcription.
- **Network Readiness Index** has an empty country column, so it contributes
  nothing at country level.
- **WJP Rule of Law, BTI** and any newer **GovTech** edition are not in the
  folder at all.

`DATA_FILES.md` lists every file, its role, its checksum, and which are unused
and why.

````python
# Which indices are currently short of evidence, and why.
from isf.engine_core import assess

for profile in assess():
    awaiting = [r["name"] for r in profile["rows"] if r["status"] == "awaiting"]
    if len(awaiting) > 6:
        print(f"{profile['index_id']:<7} {len(awaiting):>2} criteria awaiting — "
              f"{profile.get('coverage_note', 'no calculation file verified')[:60]}")
````
